```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("donnees_fors1.csv", sep=";", encoding="utf-8")

# Revenu moyen par type de ménage
revenu_menage = data.groupby("hldtyp")["idispy"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
revenu_menage.plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Revenu moyen selon le type de ménage", fontsize=16)
plt.xlabel("Type de ménage")
plt.ylabel("Revenu moyen")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


data = pd.read_csv("donnees_fors1.csv", sep=";", encoding="utf-8")

# Calculer la fréquence des niveaux d'éducation (edcat) par région
freq_niv = data.groupby(["region", "edcat"]).size().unstack(fill_value=0)

# Créer un graphique en barres empilées pour que chaque rectangle 
#represente un niveau d'education defini
freq_niv.plot(kind="bar", stacked=True, figsize=(11, 6), colormap="PiYG", edgecolor="black")


plt.title("Distribution des niveaux d'éducation par région", fontsize=16)
plt.xlabel("Régions", fontsize=12)
plt.ylabel("Fréquence des niveaux d'éducation", fontsize=12)
plt.xticks(rotation=45)
plt.legend(title="Niveaux d'éducation (edcat)")
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = pd.read_csv("donnees_fors1.csv", sep=";", encoding="utf-8")

# En lisant les donnéees j'ai vu que 3 personnes ont des revenus superieur à 500 000 francs
#selon moi  il ne devrait pas faire partie de l'echantillon donc j'ai enlevé cette valeur aberrante 
#qui pourrait compliquer l'analyse
new_data = data[data["idispy"] < 500000]

# Définir les variables à partir des données nettoyées
x = new_data["age"]  # Sélectionner la colonne âge dans les nouvelles données sans valeurs aberrantes
y = new_data["idispy"]  # Sélectionner la colonne revenu dans les nouvelles données sans valeurs aberrantes

# Calculer la régression linéaire
b, a = np.polyfit(x, y, 1) 
y_pred = b * x + a  

# Tracer le graphique et la régression linéaire
plt.scatter(x, y, color="blue", alpha=0.6, label="Données sans valeurs aberrantes")
plt.plot(x, y_pred, color="red", label="Ligne de régression")

plt.title("Distribution du revenu selon l'âge ", fontsize=16)
plt.xlabel("Âge")
plt.ylabel("Revenu")
plt.legend()
plt.show()
```

J'ai decidé de faire une regression lineaire correlant l'age et le revenu.
on voit que la regression est pas tres pentue voir presque droite même
pour la regression j'ai enlevé certaines donnéees des personne fesant partie  de l'echantillon gagne plus de 500 000 francs donc
selon moi il ne devait pas faire partie de l'echantillon donc je l'ai enlevé pour me concentrer sur la tendance generale 
De ce fait,cette regression montre que le lien entre l'age et le revenu est tres faible.
L'experience acquis au fil des années ne garantie pas un haut revenu

```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger les données
data = pd.read_csv("donnees_fors1.csv", sep=";", encoding="utf-8")

# Calculer la fréquence des type de menage par par région
grouped_data = data.groupby(["region", "hldtyp"]).size().unstack(fill_value=0)

# Créer un graphique en barres empilées
grouped_data.plot(kind="bar", stacked=True, figsize=(12, 6), colormap="plasma", edgecolor="black")

# Ajouter des détails au graphique
plt.title("Distribution de chaque type de ménage par region", fontsize=16)
plt.xlabel("Régions", fontsize=12)
plt.ylabel(" Frequences des Types de menage", fontsize=12)
plt.xticks(rotation=45)
plt.legend(title="Type de ménage")
plt.tight_layout()


plt.show()
```
